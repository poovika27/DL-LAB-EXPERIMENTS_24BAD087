```python
# EXPT NO: 2
# PART A: VISUALIZATION OF ACTIVATION FUNCTIONS
# Roll No: 24BAD087
# Name: POOVIKA M

import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 200)

# Activation functions
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

# Plot activation functions
plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid, label='Sigmoid', linewidth=2)
plt.plot(x, tanh, label='Tanh', linewidth=2)
plt.plot(x, relu, label='ReLU', linewidth=2)

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()

# Comparison
print("Activation Function Comparison")
print("Sigmoid : Output Range = (0,1)")
print("Tanh    : Output Range = (-1,1)")
print("ReLU    : Output Range = (0,∞)")
```

```python
# EXPT NO: 2
# PART B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS
# Roll No: 24BAD087
# Name: POOVIKA M

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

activation_functions = ['sigmoid', 'tanh', 'relu']
history_dict = {}

for activation in activation_functions:
    print("\nTraining ANN using", activation.upper(), "activation function")

    model = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation=activation),
        Dense(64, activation=activation),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        x_train,
        y_train,
        epochs=10,
        validation_split=0.2,
        verbose=1
    )

    test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

    history_dict[activation] = history.history

    print("Training Accuracy :", history.history['accuracy'][-1])
    print("Validation Accuracy :", history.history['val_accuracy'][-1])
    print("Training Loss :", history.history['loss'][-1])
    print("Validation Loss :", history.history['val_loss'][-1])
    print("Test Accuracy :", test_accuracy)


# Accuracy Graph
plt.figure(figsize=(8,5))

for activation in activation_functions:
    plt.plot(history_dict[activation]['val_accuracy'], label=activation)

plt.title("Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# Loss Graph
plt.figure(figsize=(8,5))

for activation in activation_functions:
    plt.plot(history_dict[activation]['val_loss'], label=activation)

plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# EXPT NO: 2
# PART C: COMPARISON OF OPTIMIZATION ALGORITHMS
# Roll No: 24BAD087
# Name: POOVIKA M

import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD, RMSprop, Adam

# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

optimizers = {
    "SGD": SGD(),
    "Momentum": SGD(momentum=0.9),
    "RMSProp": RMSprop(),
    "Adam": Adam()
}

history_dict = {}

for name, optimizer in optimizers.items():

    print("\nTraining using", name)

    model = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        x_train,
        y_train,
        epochs=10,
        validation_split=0.2,
        verbose=1
    )

    test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

    history_dict[name] = history.history

    print("Training Accuracy :", history.history['accuracy'][-1])
    print("Validation Accuracy :", history.history['val_accuracy'][-1])
    print("Training Loss :", history.history['loss'][-1])
    print("Validation Loss :", history.history['val_loss'][-1])
    print("Test Accuracy :", test_accuracy)


# Accuracy Graph
plt.figure(figsize=(8,5))

for name in optimizers.keys():
    plt.plot(history_dict[name]['val_accuracy'], label=name)

plt.title("Optimizer Comparison - Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()


# Loss Graph
plt.figure(figsize=(8,5))

for name in optimizers.keys():
    plt.plot(history_dict[name]['val_loss'], label=name)

plt.title("Optimizer Comparison - Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```
